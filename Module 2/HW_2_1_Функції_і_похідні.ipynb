1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return 10 * x - 5

def df(func, x, h=1e-6):
    return (func(x + h) - func(x)) / h

xs = np.linspace(0, 10, 1000)

plt.figure(figsize=(10, 5))
plt.plot(xs, f(xs), label='f(x)', color='blue')
plt.plot(xs, df(f, xs), linestyle='--', color='green',label="f'(x)")

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

Аналізуючм графік, бачимо, що зі збільшенням кількості годин зростає і оцінка (+10 балів за кожну наступну годину)

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
import sympy as sp

x = sp.symbols('x')

f1 = sp.sin(x)
df1 = sp.diff(f1, x)
print(f"Функція 1: {f1}, її похідна: {df1}")

xs = np.linspace(-10, 10, 1000)

plt.figure(figsize=(10, 5))
plt.plot(xs, np.sin(xs), label='f(x)=sin(x)', color='blue')
plt.plot(xs, np.cos(xs), linestyle='--', color='green',label="f'(x) = cos(x)")

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
x = sp.symbols('x')

f2 = sp.ln(x)
df2 = sp.diff(f2, x)
print(f"Функція 2: {f2}, її похідна: {df2}")

xs = np.linspace(0.01, 1.5, 1000)

plt.figure(figsize=(10, 5))
plt.plot(xs, np.log(xs), label='f(x)=ln(x)', color='blue')
plt.plot(xs, 1/xs, linestyle='--', color='green',label="f'(x) = 1/x")

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')

f = x**2 - 5*x + 1
df = sp.diff(f, x)
print(f"Функція 2: {f}, її похідна: {df}")

xs = np.linspace(0, 100, 1000)

plt.figure(figsize=(10, 5))
plt.plot(xs, xs**2 - 5*xs + 1, label='f(x)=x^2 - 5x + 1', color='blue')
plt.plot(xs, 2*xs - 5, linestyle='--', color='green',label="f'(x) = 2x")

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```
